## Backpropagation: com aprèn la xarxa

Fins ara la xarxa fa prediccions amb pesos **aleatoris**. Backpropagation
és l'algorisme que calcula com s'ha de canviar cada pes perquè la
predicció s'apropi a la resposta correcta.

### Pas 1: Mesurar l'error (funció de pèrdua)

Fem servir l'error quadràtic mitjà:

$$L = \frac{1}{2}(y - A^{[2]})^2$$

On $y$ és la resposta correcta i $A^{[2]}$ la predicció de la xarxa.

### Pas 2: La derivada de la sigmoid

Per saber com canvia la sortida d'una neurona quan canvia el seu pes,
necessitem la derivada de $\sigma$. Té una propietat molt còmoda:

$$\sigma'(x) = \sigma(x) \cdot (1 - \sigma(x))$$

### Pas 3: Regla de la cadena (cap enrere, capa per capa)

**Capa de sortida:**

$$dZ^{[2]} = (A^{[2]} - y) \cdot \sigma'(Z^{[2]})$$

$$dW^{[2]} = A^{[1]T} \cdot dZ^{[2]} \qquad db^{[2]} = dZ^{[2]}$$

**Capa oculta** (l'error es "propaga cap enrere" a través de $W^{[2]}$):

$$dA^{[1]} = dZ^{[2]} \cdot W^{[2]T}$$

$$dZ^{[1]} = dA^{[1]} \cdot \sigma'(Z^{[1]})$$

$$dW^{[1]} = X^{T} \cdot dZ^{[1]} \qquad db^{[1]} = dZ^{[1]}$$

### Pas 4: Actualitzar els pesos (gradient descent)

$$W = W - \alpha \cdot dW \qquad b = b - \alpha \cdot db$$

On $\alpha$ (alpha) és la **taxa d'aprenentatge**: com de grans són els passos.

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivada(a):
    # 'a' ja és la sortida de sigmoid, no l'entrada original
    return a * (1 - a)

def forward_pass(X, W1, b1, W2, b2):
    Z1 = np.dot(X, W1) + b1
    A1 = sigmoid(Z1)
    Z2 = np.dot(A1, W2) + b2
    A2 = sigmoid(Z2)
    return A1, A2

def backward_pass(X, y, A1, A2, W2):
    # Capa de sortida
    dZ2 = (A2 - y) * sigmoid_derivada(A2)
    dW2 = np.outer(A1, dZ2)
    db2 = dZ2

    # Capa oculta
    dA1 = np.dot(dZ2, W2.T)
    dZ1 = dA1 * sigmoid_derivada(A1)
    dW1 = np.outer(X, dZ1)
    db1 = dZ1

    return dW1, db1, dW2, db2
```

```python
# Prova: un pas d'entrenament

X = np.array([0.5, 0.3, 0.8])
y = np.array([1.0, 0.0])          # resposta correcta esperada

W1 = np.random.randn(3, 4)
b1 = np.random.randn(4)
W2 = np.random.randn(4, 2)
b2 = np.random.randn(2)

alpha = 0.5   # taxa d'aprenentatge

# Abans d'entrenar
A1, A2 = forward_pass(X, W1, b1, W2, b2)
error_abans = np.mean((y - A2) ** 2)

# Backpropagation
dW1, db1, dW2, db2 = backward_pass(X, y, A1, A2, W2)

# Actualitzar pesos
W1 -= alpha * dW1
b1 -= alpha * db1
W2 -= alpha * dW2
b2 -= alpha * db2

# Després d'un pas d'entrenament
A1, A2 = forward_pass(X, W1, b1, W2, b2)
error_despres = np.mean((y - A2) ** 2)

print(f"Predicció abans:   {A2 if False else None}")  # placeholder
print(f"Error abans:   {error_abans:.5f}")
print(f"Error després: {error_despres:.5f}")
print(f"L'error ha {'disminuït ✓' if error_despres < error_abans else 'augmentat ✗'}")
```
