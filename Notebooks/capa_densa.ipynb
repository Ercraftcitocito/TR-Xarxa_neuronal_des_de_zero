## Capa densa

Una capa densa és un conjunt de neurones que reben **totes** les entrades
i cada una genera una sortida independent.

Si tenim $n$ entrades i $m$ neurones, la capa fa:

$$Z = X \cdot W + b$$

On:
- $X$ és el vector d'entrades, de forma $(1 \times n)$
- $W$ és la matriu de pesos, de forma $(n \times m)$
- $b$ és el vector de biaixos, de forma $(1 \times m)$
- $Z$ és la sortida abans de l'activació, de forma $(1 \times m)$

La sortida final de la capa és:

$$A = \sigma(Z)$$

Cada columna de $W$ representa els pesos d'una neurona concreta.

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def capa_densa(X, W, b):
    Z = np.dot(X, W) + b       # combinació lineal: (1×n) · (n×m) = (1×m)
    return sigmoid(Z)          # activació de cada neurona

# Exemple: 3 entrades, 4 neurones
X = np.array([0.5, 0.3, 0.8])  # vector d'entrades: 3 entrades
W = np.random.randn(3, 4)      # matriu de pesos: 3 files | 4 columnes
b = np.random.randn(4)         # vector de biaixos: 4 biaixos per a cada neurona

A = capa_densa(X, W, b)

print(W)
print(b)
print(f"Sortida de la capa (4 neurones): {A}")
```

```python
# Visualització dels pesos de cada neurona
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 4, figsize=(12, 3))

for i, ax in enumerate(axes):
    ax.bar(range(3), W[:, i], color='steelblue', alpha=0.7)
    ax.axhline(0, color='gray', linewidth=0.8)
    ax.set_title(f'Neurona {i+1}')
    ax.set_xlabel('Entrada')
    ax.set_ylabel('Pes')
    ax.set_xticks(range(3))
    ax.set_xticklabels(['x₁', 'x₂', 'x₃'])

plt.suptitle('Pesos de cada neurona de la capa', fontsize=13)
plt.tight_layout()
plt.show()
```
